```python
def greet(name: str) -> str:
  """Возвращает приветствие для указанного имени."""
  return f"Привет, {name}!"

def compute_statistics(numbers):
  """Считает базовые статистики по списку чисел."""
  if not numbers:
    return None
  return {
    "count": len(numbers),
    "sum": sum(numbers),
    "min": min(numbers),
    "max": max(numbers),
    "avg": sum(numbers) / len(numbers),
    }
```

```python
def factorial(n: int) -> int:
    """Рекурсивный факториал. n >= 0."""
    if n < 0:
        raise ValueError("factorial не определён для отрицательных чисел")
    return 1 if n <= 1 else n * factorial(n - 1)
```

```python
print(f"5! = {factorial(5)}")
print(f"10! = {factorial(10)}")
```

```python
# 1. Приветствие
print(greet("Студент"))
# 2. Работа со списком
sample = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]
stats = compute_statistics(sample)
print(f"Список: {sample}")
for key, value in stats.items():
    print(f" {key:>5}: {value}")
```
