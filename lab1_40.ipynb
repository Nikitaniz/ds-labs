```python
def is_leap_year(y: int) -> bool:

    #Проверяем, является ли год високосным. Год високосный, если: делится на 4 (потому что високосный год каждые 4 года),
    #но не делится на 100(так как разница за 100 лет одни лишние сутки чтобы их убрать висакосный год пропускают),
    #или делится на 400(за 400 лет теряются одни сутки поэтому раз в 400 лет висакосный код пропускают).

    return (y % 4 == 0 and y % 100 != 0) or (y % 400 == 0)


def day_of_year(d: int, m: int, y: int) -> int:

    #Возвращает порядковый номер дня в году.
    #Алгоритм:
    # 1. Определяем, високосный ли год (февраль: 28 или 29 дней).
    # 2. Формируем список длин месяцев для данного года.
    #  3. Складываем количество дней во всех месяцах до m-го
    #     и прибавляем текущий день d.

    # Список длин месяцев для невисокосного года
    days_in_months = [31, 28, 31, 30, 31, 30,
                      31, 31, 30, 31, 30, 31]

    # Корректируем февраль для високосного года
    if is_leap_year(y):
        days_in_months[1] = 29

    # Сумма дней в месяцах до текущего + текущий день
    result = sum(days_in_months[:m - 1]) + d
    return result
```
